```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style and palette
sns.set_theme(style="whitegrid")
palette = sns.color_palette("muted")

# Load dataset
df = pd.read_csv("codealpha_scraped_books.csv")

# Clean price column if not already done
df['Price_Clean'] = df['Price'].str.replace('£', '', regex=False).astype(float)

# Figure: Multi-plot Advanced Dashboard
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# 1. Price vs Star Rating (if rating column exists, or custom distribution)
sns.histplot(df['Price_Clean'], kde=True, ax=axes[0], color='teal', bins=15)
axes[0].set_title('Price Frequency Distribution', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Price (£)', fontsize=12)
axes[0].set_ylabel('Count', fontsize=12)

# 2. Cumulative Price Distribution
sns.ecdfplot(data=df, x='Price_Clean', ax=axes[1], color='purple', linewidth=2)
axes[1].set_title('Cumulative Distribution of Book Prices', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Price (£)', fontsize=12)
axes[1].set_ylabel('Cumulative Probability', fontsize=12)

plt.tight_layout()
plt.show()
```
